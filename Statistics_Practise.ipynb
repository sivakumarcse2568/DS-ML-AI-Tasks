```python
#Measure of Central Tendency
#1. MEAN

import numpy as np
ages = [22,24,28,36,22,2,26]
print("Mean of Ages: ",np.mean(ages))
```

```python
#2.MEDIAN

print("Median of Ages: ",np.median(ages))
```

```python
#3.MODE

from scipy.stats import mode
print("Mode(Frequent Value): ",mode(ages)[0])
```

```python
#Measures of Dispersion
#1.Range

range = max(ages) - min(ages)
print("Range: ",range)
```

```python
#2.InterQuartile Range

import pandas as pd
series = pd.Series(ages)
q1 = series.quantile(0.25)
q3 = series.quantile(0.75)
ir = q3 - q1
print("Q1: ",q1)
print("Q3: ",q3)
print("IQR: ",ir)
```

```python
#3. Standard Deviation & Variation

print("Standard Deviation: ",np.std(ages))
print("Variance: ",np.var(ages))
```

```python
#Measures of Frequency
#1.Kurtosis

print("Kurtosis: ",series.kurt())
```

```python
#2.Skewness

print("Skewness: ",series.skew())
```
