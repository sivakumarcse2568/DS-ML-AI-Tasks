```python
import numpy as np
arr = np.array([10,20,30,40,50])
print("Array", arr)
```

```python
arr1 = np.arange(1,21)
print("Array (1-20)",arr1)
```

```python
arr2 = np.zeros(10)
print("Array of 10 Zeroes:",arr2)
```

```python
arr3 = np.ones((3,4))
print("3x4 Array of Ones:\n",arr3)
```

```python
arr4 = np.full((2,3),7)
print("2x3 Array with 7:\n",arr4)
```

```python
arr5 = np.linspace(0,100,10)
print("10 Equally spaced Array:\n",arr5)
```

```python
arr6 = np.eye(4,4)
print("4x4 Identity Matrix:\n",arr6)
```

```python
arr7 = np.arange(2,51,2)
print("Array of Even Numbers(2-50):\n",arr7)
```

```python
a = np.arange(1,13)
print("Reshape (3x4)\n",a.reshape(3,4))
```

```python
print("Reshape (4x3)\n",a.reshape(4,3))
```

```python
print("Reshape (2x2x3)\n",a.reshape(2,2,3))
```
