```python
import numpy as np
print("Even:",np.arange(22, 71, 2))
```

```python
print("Odd:",np.arange(21, 72, 2))
```

```python
arr = np.array([10, 20, 30, 40, 50])
arr = arr.astype(float)
print("Int to Float:",arr)
```

```python
import numpy as np
a = np.arange(10)
print("Sliced arr:",a[2:5])
```

```python
x = np.array([-5, 9, 20, 25, -3, 5, 16, 10])
x[(x >= -5) & (x <= 15)] *= -1

print(x)
```

```python
arr = np.array([0, 1, 2, 3, 4, 5])
print("Dimension:",arr.ndim)
print("Shape:",arr.shape)
```

```python
arr = np.array([0, 1, 2, 3, 4, 5])
print("First Elem:",arr[0])
print("Last Elem:",arr[-1])
```

```python
start, length, step = map(int, input().split())
arr = np.arange(start, start + length * step, step)
print(arr)
```

```python
start, end, step = map(float, input().split())
arr = np.arange(start, end, step)
print(np.round(arr, 2))
```
