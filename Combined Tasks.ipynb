```python
import numpy as np
import pandas as pd
data = pd.read_csv("/content/insurance.csv")
```

```python
#NUMPY
#1. Convert the age column into a NumPy array.
age = np.array(data["age"])
print(age)
```

```python
#2.Convert the BMI column into a NumPy array.
bmi = np.array(data["bmi"])
print(bmi)
```

```python
#3.Extract the insurance charges into an array.
charges = np.array(data["charges"])
print(charges)
```

```python
#4.Find the total number of customers.
print("Total Customers:",len(age))
```

```python
#5.Find the average age
print("Average Age:",np.mean(age))
```

```python
#6.Find minimum and maximum BMI
print("Minimum BMI:",np.min(bmi))
print("Maximum BMI:",np.max(bmi))
```

```python
#7.Calculate average insurance charges
print("Average Charges:",np.mean(charges))
```

```python
#8.Find median insurance charges
print("Median Charges:",np.median(charges))
```

```python
#9.Calculate standard deviation of charges
print("Standard Deviation:",np.std(charges))
```

```python
#10.Count customers above 40 years old
count = np.sum(age > 40)
print("Customers above 40:",count)
```

```python
#Level 2 – Numerical Analysis
#Find customers whose insurance charges are greater than $20,000
high_charges = charges[charges > 20000]
print(high_charges)
```

```python
#Find customers whose BMI is between 25 and 30
filtered_bmi = bmi[(bmi >= 25)&(bmi <= 30)]
print(filtered_bmi)
```

```python
#Calculate 25th, 50th and 75th percentiles
print("25th Percentile:",np.percentile(charges,25))
print("50th Percentile:",np.percentile(charges,50))
print("75th Percentile:",np.percentile(charges,75))
```

```python
#Calculate variance of insurance charges
print("Variance:", np.var(charges))
```

```python
#Normalize the age array using Min-Max normalization
normalized_age = (age-np.min(age))/(np.max(age)-np.min(age))
print(normalized_age)
```

```python
#Calculate the difference between maximum and minimum charges
difference = np.max(charges)-np.min(charges)
print("Difference:", difference)
```

```python
#Calculate average charges by gender and smoker status
sex = np.array(data["sex"])
smoker = np.array(data["smoker"])
print("Average Charges by Gender")
for gender in np.unique(sex):
    print(gender, np.mean(charges[sex == gender]))
print("Average Charges by Smoker Status")

for status in np.unique(smoker):
    print(status, np.mean(charges[smoker == status]))
```

```python
#Find the region with the highest average insurance charges
region = np.array(data["region"])
for location in np.unique(region):
    average = np.mean(charges[region == location])
    print(location, average)
```

```python
#Compare average charges of smokers and non-smokers
smoker_charges = np.mean(charges[smoker == "yes"])
non_smoker_charges = np.mean(charges[smoker == "no"])
print("Smoker Average Charges:", smoker_charges)
print("Non-Smoker Average Charges:", non_smoker_charges)
```

```python
#Find average charges for each combination of region and gender
for location in np.unique(region):
    for gender in np.unique(sex):
        average = np.mean(charges[(region == location) & (sex == gender)])
        print(location, gender, average)
```

```python
#Level 2 – Filtering and Transformation
#Extract customers aged above 50
result = data[data["age"] > 50]
print(result)
```

```python
#Filter customers with BMI above 30
result = data[data["bmi"] > 30]
print(result)
```

```python
#Extract female smokers
result = data[(data["sex"] == "female") & (data["smoker"] == "yes")]
print(result)
```

```python
#Find customers from the southeast region
result = data[data["region"] == "southeast"]
print(result)
```

```python
#Filter charges between $10,000 and $30,000
result = data[(data["charges"] >= 10000) & (data["charges"] <= 30000)]
print(result)
```

```python
#Create a new column called age_group
data["age_group"] = pd.cut(data["age"],bins=[0, 20, 40, 60, 100],labels=["Young", "Adult", "Middle Age", "Senior"])
print(data[["age", "age_group"]])
```

```python
#Create a new column called bmi_category
data["bmi_category"] = pd.cut(data["bmi"],bins=[0, 18.5, 25, 30, 100],labels=["Underweight", "Normal", "Overweight", "Obese"])
print(data[["bmi", "bmi_category"]])
```

```python
#Convert smoker values into numerical values
data["smoker_number"] = data["smoker"].map({"yes": 1,"no": 0})
print(data[["smoker", "smoker_number"]])
```

```python
#Create a new column called high_cost_customer
data["high_cost_customer"] = data["charges"] > 20000
print(data[["charges", "high_cost_customer"]])
```

```python
#Level 3 – Grouping and Aggregation
#Calculate average charges by gender
result = data.groupby("sex")["charges"].mean()
print(result)
```

```python
#Find total charges by region
result = data.groupby("region")["charges"].sum()
print(result)
```
