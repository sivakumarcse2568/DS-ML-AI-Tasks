```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv("/content/netflix_titles.csv")
```

```python
#1) Display the total number of Movies and TV Shows using a bar chart.
count = data['type'].value_counts()
print(count)
```

```python
plt.bar(count.index,count.values,edgecolor='blue',color='yellow',linewidth=2)
plt.title("Movies VS TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Total Count")
```

```python
#2) Create a pie chart showing content percentages.
plt.pie(count.values,labels=count.index,autopct='%1.2f%%')
plt.title("Movies and TV Shows Percentage")
plt.legend()
plt.show()
```

```python
#3) Create a histogram for movie durations.
movies = data[data['type'] == 'Movie']
duration = movies['duration'].str.replace(' min', '',regex=False)
duration = duration.dropna().astype(int)
plt.hist(duration,bins=10,edgecolor='blue',color='yellow',linewidth=2)
plt.title("Movie Duration Frequency")
plt.xlabel("Duration")
plt.ylabel("No of Movies")
plt.show()
```

```python
#5) Create a line chart showing content released per year.
year = data['release_year'].value_counts().sort_index()
plt.plot(year.index,year.values,color='yellow',marker='o',markeredgecolor='blue')
plt.title("Content Released Per Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Contents")
plt.show()
```

```python
#6) Compare Movies and TV Shows using two subplots.
movies = data[data['type'] == 'Movie']
shows = data[data['type'] == 'TV Show']
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.hist(movies['release_year'],bins=10,edgecolor='blue',color='yellow')
plt.title("Movies Released")
plt.xlabel("Year")
plt.ylabel("Count")

plt.subplot(1, 2, 2)
plt.hist(shows['release_year'],bins=10,edgecolor='blue',color='yellow')
plt.title("TV Shows Released")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

```python
#7) Find the top 10 most common content ratings and visualize them.
rating = data['rating'].value_counts().head(10)
plt.bar(rating.index,rating.values,edgecolor='blue',color='yellow',linewidth=2)
plt.title("Top 10 Content Ratings")
plt.xlabel("Ratings")
plt.ylabel("Count")
plt.show()
```

```python
#8) Create a box plot to identify movie duration outliers.
movies = data[data['type'] == 'Movie']
duration = movies['duration'].str.replace(' min','',regex=False)
duration = duration.dropna().astype(int)
plt.boxplot(duration)
plt.title("Movie Duration Outliers")
plt.ylabel("Duration")
plt.show()
```

```python
#9) Compare the number of Movies and TV Shows released each year using two lines.
movies = data[data['type'] == 'Movie']
shows = data[data['type'] == 'TV Show']
movie_count = movies['release_year'].value_counts().sort_index()
show_count = shows['release_year'].value_counts().sort_index()
plt.plot(movie_count.index, movie_count.values, label='Movies',color='yellow',marker='o',markeredgecolor='blue')
plt.plot(show_count.index, show_count.values, label='TV Shows',color='red',marker='d',markeredgecolor='black')
plt.title("Movies vs TV Shows Released Each Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Releases")
plt.legend()
plt.show()
```
