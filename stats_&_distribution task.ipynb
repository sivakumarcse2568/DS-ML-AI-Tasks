```python
import pandas as pd
data = pd.read_csv("/content/insurance.csv")
```

```python
#LEVEL-1
print("AGE")
print("Mean:",data['age'].mean())
print("Median:",data['age'].median())
print("Mode:",data['age'].mode()[0])
print("Standard Deviation:",data['age'].std())

print("\nBMI")
print("Mean:",data['bmi'].mean())
print("Median:",data['bmi'].median())
print("Mode:",data['bmi'].mode()[0])
print("Standard Deviation:",data['bmi'].std())

print("\nCHARGES")
print("Mean:",data['charges'].mean())
print("Median:",data['charges'].median())
print("Mode:",data['charges'].mode()[0])
print("Standard Deviation:",data['charges'].std())
```

```python
#Plot Histograms for All Three Numerical Columns
#i) Age Histogram
import matplotlib.pyplot as plt
plt.hist(data['age'],bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#ii) BMI Histogram
plt.hist(data['bmi'],bins=10)
plt.title("BMI Distribution")
plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.show()
```

```python
#Charges Histogram
plt.hist(data['charges'],bins=20)
plt.title("Insurance Charges Distribution")
plt.xlabel("Charges")
plt.ylabel("Frequency")
plt.show()
```

```python
#Identify Which Columns Appear Normally Distributed
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.hist(data['age'],bins=10)
plt.title("Age")

plt.subplot(1,3,2)
plt.hist(data['bmi'],bins=15)
plt.title("BMI")

plt.subplot(1,3,3)
plt.hist(data['charges'],bins=20)
plt.title("Charges")
plt.show()
```

```python
#Calculate Skewness and Kurtosis
print("AGE")
print("Skewness:",data['age'].skew())
print("Kurtosis:",data['age'].kurt())
print()
print("BMI")
print("Skewness:",data['bmi'].skew())
print("Kurtosis:",data['bmi'].kurt())
print()
print("CHARGES")
print("Skewness:",data['charges'].skew())
print("Kurtosis:",data['charges'].kurt())
```

```python
#Identify Whether Charges Are Positively or Negatively Skewed
skewness = data['charges'].skew()
print("Charges Skewness:",skewness)
if skewness > 0:
    print("Charges are positively skewed")
else:
    print("Charges are negatively skewed")
```

```python
#Create Boxplots and Identify Outliers
plt.boxplot(data['charges'])
plt.title("Insurance Charges Boxplot")
plt.ylabel("Charges")
plt.show()
```

```python
#Compare Mean and Median.
print("Charges Mean:",data['charges'].mean())
print("Charges Median:",data['charges'].median())
```

```python
#LEVEL 2
#Generate 1,000 Random Samples from a Normal Distribution
import numpy as np
normal_data = np.random.normal(50, 10, 1000)
print(normal_data)
```

```python
#Plot a Normal Distribution Curve
plt.hist(normal_data, bins=30)
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Density")
plt.show()
```

```python
#Calculate Probability That a Customer Has BMI Above 30
from scipy import stats
mean = data['bmi'].mean()
std = data['bmi'].std()
probability = 1 - stats.norm.cdf(30,mean,std)
print("Theoretical Probability:",probability)
```

```python
#Generate a Binomial Distribution Using Smoker Proportion
p = (data['smoker'] == 'yes').mean()
print("Smoker Probability:",p)
n = 10
x = np.arange(0, 11)
prob = stats.binom.pmf(x, n, p)
plt.bar(x, prob)
plt.title("Binomial Distribution")
plt.xlabel("Number of Smokers")
plt.ylabel("Probability")
plt.show()
```

```python
#Calculate Probability of Getting Exactly 5 Smokers in 10 Customers
p = (data['smoker'] == 'yes').mean()
probability = stats.binom.pmf(5, 10, p)
print("Probability of Exactly 5 Smokers:", probability)
```

```python
#Generate a Poisson Distribution and Visualize It
x = np.arange(0, 11)
poisson = stats.poisson.pmf(x, 3)
plt.bar(x, poisson)
plt.title("Poisson Distribution")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.show()
```

```python
uniform = np.random.uniform(0, 100, 1000)
normal = np.random.normal(50, 15, 1000)
plt.hist(uniform,bins=20,label='Uniform')
plt.hist(normal,bins=20,label='Normal')
plt.title("Uniform vs Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```
